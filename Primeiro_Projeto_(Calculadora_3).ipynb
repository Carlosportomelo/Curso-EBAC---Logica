```python
nome = input('Olá! Digite seu nome para avançar: ')
print('Ola, '+ nome + '! Seja bem vindo :)')

while True:
  valor_da_compra_str = input('Agora, digite o valor da sua compra (ou digite "sair" para finalizar): ')
  if valor_da_compra_str.lower() == 'sair':
    break
  try:
    valor_da_compra = int(valor_da_compra_str)
    if valor_da_compra > 200:
      print ('Desconto de 20%')
      print ('R$'+ str(valor_da_compra * 0.8))
    elif valor_da_compra > 150:
      print ('Desconto de 10%')
      print ('R$'+ str(valor_da_compra * 0.9))
    else:
      print('Sem desconto')
  except ValueError:
    print('Entrada inválida. Por favor, digite um número ou "sair".')

print('Programa finalizado. Obrigado!')
```
